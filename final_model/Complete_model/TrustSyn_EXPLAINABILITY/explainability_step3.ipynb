```python
import os

search_dir = "/Users/konuri/stacking/TrustSyn_DMPNN/FINAL_MODEL"

keywords = [
    "attention",
    "atom_weights",
    "edge_weights",
    "gradient",
    "integrated",
    "saliency"
]

for root, dirs, files in os.walk(search_dir):
    for file in files:
        if file.endswith(".ipynb"):
            path = os.path.join(root,file)

            text = open(path, errors="ignore").read()

            found = [
                k for k in keywords
                if k.lower() in text.lower()
            ]

            if found:
                print(
                    file,
                    "->",
                    found
                )
```
