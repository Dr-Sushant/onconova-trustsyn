```python
# ============================================================
# TrustSyn TRUST LAYER v3.1
# Conformal calibration from meta predictions
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path


BASE = Path("/Users/konuri/stacking/STACKING_TABLES")


FILES = {
    "RANDOM":
        BASE / "RANDOM_META_PREDICTIONS.csv",

    "COLD_COMBINATION":
        BASE / "COLD_COMBINATION_META_PREDICTIONS.csv",

    "COLD_CELL":
        BASE / "COLD_CELL_META_PREDICTIONS.csv"
}


errors = []


for split, file in FILES.items():

    print("\n================")
    print(split)

    df = pd.read_csv(file)

    print("Loaded:", df.shape)


    temp_error = np.abs(
        df["y_true"] -
        df["meta_prediction"]
    )


    errors.extend(
        temp_error.tolist()
    )


    print(
        "Mean absolute error:",
        temp_error.mean()
    )

    print(
        "Samples:",
        len(temp_error)
    )



errors = np.array(errors)


# 90% conformal calibration error
calibration_error = np.quantile(
    errors,
    0.90
)


print("\n==============================")
print(
    "90% CALIBRATION ERROR:",
    calibration_error
)


out = pd.DataFrame(
    {
        "metric":
        [
            "90_percent_absolute_error"
        ],

        "value":
        [
            calibration_error
        ]
    }
)


save = BASE / "TRUST_LAYER_v3_1_CALIBRATION.csv"


out.to_csv(
    save,
    index=False
)


print("\nSAVED:")
print(save)
```
